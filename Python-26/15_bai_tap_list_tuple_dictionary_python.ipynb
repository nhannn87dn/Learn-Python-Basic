# Bài 1 — Truy cập và thay đổi phần tử trong List

### Yêu cầu
Cho một danh sách các số nguyên. Hãy:
1. In ra phần tử đầu tiên.
2. In ra phần tử cuối cùng.
3. Thay đổi phần tử ở vị trí thứ 2 thành `100`.
4. In ra danh sách sau khi thay đổi.

### Biến đầu vào
- `numbers`: `list[int]`
- Danh sách có ít nhất 3 phần tử.

### Kết quả đầu ra
- In phần tử đầu tiên.
- In phần tử cuối cùng.
- In danh sách sau khi cập nhật.

### Ví dụ
**Input**
```python
numbers = [10, 20, 30, 40, 50]
```

**Output**
```text
10
50
[10, 100, 30, 40, 50]
```

# Bài 15 — Quản lý điểm sinh viên bằng List, Tuple và Dictionary

### Yêu cầu
Cho Dictionary chứa tên sinh viên và Tuple điểm `(Toán, Văn, Anh)`.
Hãy:
1. Tính điểm trung bình cho từng sinh viên.
2. Tạo Dictionary mới có dạng `{"tên": điểm_trung_bình}`.
3. Tìm sinh viên có điểm trung bình cao nhất.

### Biến đầu vào
- `students`: `dict[str, tuple[float, float, float]]`
- Mỗi Tuple gồm đúng 3 điểm.

### Kết quả đầu ra
- `averages`: Dictionary chứa điểm trung bình của từng sinh viên.
- Tên sinh viên có điểm trung bình cao nhất.
- Điểm trung bình cao nhất, làm tròn 2 chữ số thập phân.

### Ví dụ
**Input**
```python
students = {
    "An": (8, 7, 9),
    "Bình": (9, 8, 8),
    "Cường": (7, 6, 8)
}
```

**Output**
```text
averages = {
    "An": 8.0,
    "Bình": 8.33,
    "Cường": 7.0
}

Bình
8.33
```
